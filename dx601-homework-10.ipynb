### Problem 1

The vector $$\mathbf{x}=\begin{bmatrix} 5 \\ 1 \\ 4 \end{bmatrix}$$ is generated by $$\mathbf{g}_1=\begin{bmatrix} 1 \\ 0 \\ 1 \end{bmatrix}, \mathbf{g}_2 = \begin{bmatrix} 0 \\ 1 \\ -1 \end{bmatrix}.$$
Set `p1` to $(a,b)$ where $\mathbf{x}=a \mathbf{g}_1 + b \mathbf{g}_2$.

```python
# YOUR CHANGES HERE

# Define vectors
x = np.array([5, 1, 4])
g1 = np.array([1, 0, 1])
g2 = np.array([0, 1, -1])


# Combine the vectors as vertical columns
A = np.column_stack((g1, g2))

# Solve the linear system for unknowns a and b
a, b = np.linalg.lstsq(A, x, rcond=None)[0]

# Store the rounded coordinates as a tuple (a, b)
p1 = (round(a), round(b))
```

### Problem 5

Let $$B = \left\{ \begin{bmatrix} 2 \\ 1 \end{bmatrix}, \begin{bmatrix} 1 \\ 2 \end{bmatrix} \right\}.$$
For $$\mathbf{x} = \begin{bmatrix} 7 \\ 8 \end{bmatrix},$$ find $[x]_B$, the coordinates of $\mathbf{x}$ with respect to $B$.
Set `p5` to $[x]_B$.

```python
# Define vectors 
b1 = np.array([2, 1])
b2 = np.array([1, 2])
x = np.array([7, 8])

# Combine b1 and b2 to form the basis
B = np.column_stack((b1, b2))

# Solve B * [x]B = x
p5 = np.linalg.solve(B, x)
```
